```python
# write programm to count no of vowel in string
name = input("enter your string")
vowel = "AEIOUaeiou"
count = 0


for i in name:
    if i in vowel:
        count+=1
        
    elif i not in vowel :
        count+=1
      
    else :
        print("error")
        
print("vowel",count)
print("consonant",count)
```

```python
# check no is palindrome

num = "129"
if (num) == (num[::-1]):
    print("number is palindrome")

else :
    print("not palindrome number")
```
